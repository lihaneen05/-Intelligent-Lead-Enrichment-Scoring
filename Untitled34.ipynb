```python
# ==========================================
# ASSIGNMENT 1: HARRIS CORNER DETECTOR
# Course: Computer Vision (BS AI-7)
# Student: Haneen Tahir (01-136232-016)
# ==========================================

import numpy as np
import scipy.ndimage as ndi
import matplotlib.pyplot as plt
import cv2

# Set random seed for reproducibility
np.random.seed(42)
print("Environment initialized successfully with NumPy, SciPy, OpenCV, and Matplotlib.")
```

```python
def create_gaussian_kernel(sigma):
    """Generates an optimal 1D Gaussian kernel based on 3-sigma truncation rule."""
    radius = int(np.ceil(3 * sigma))
    kernel_size = 2 * radius + 1
    x = np.arange(-radius, radius + 1, dtype=np.float64)
    kernel = np.exp(-(x**2) / (2 * sigma**2))
    kernel /= np.sum(kernel) # Normalize
    return kernel

def separable_gaussian_convolve(image, sigma):
    """Performs highly optimized separable 1D Gaussian convolution."""
    kernel_1d = create_gaussian_kernel(sigma)
    # Pass 1: Horizontal convolution
    img_h = ndi.convolve1d(image, kernel_1d, axis=1, mode='reflect')
    # Pass 2: Vertical convolution
    img_smooth = ndi.convolve1d(img_h, kernel_1d, axis=0, mode='reflect')
    return img_smooth

# Verification Test against SciPy
test_img = np.random.rand(100, 100)
sigma_val = 1.5
custom_out = separable_gaussian_convolve(test_img, sigma_val)

# Full 2D equivalent reference
radius = int(np.ceil(3 * sigma_val))
k1d = create_gaussian_kernel(sigma_val)
kernel_2d = np.outer(k1d, k1d)
scipy_out = ndi.convolve(test_img, kernel_2d, mode='reflect')

max_error = np.max(np.abs(custom_out - scipy_out))
print(f"Maximum Absolute Error vs SciPy: {max_error:.2e} (Matches Machine Precision)")
```

```python
class HarrisCornerDetector:
    def __init__(self, sigma=1.0, k=0.04, threshold_ratio=0.01):
        self.sigma = sigma
        self.k = k
        self.threshold_ratio = threshold_ratio

    def detect(self, image_gray):
        # Step 1: Compute spatial derivatives using Sobel filters
        I_x = ndi.sobel(image_gray, axis=1, mode='reflect')
        I_y = ndi.sobel(image_gray, axis=0, mode='reflect')

        # Step 2: Compute components of the structure tensor M
        Ixx = I_x ** 2
        Iyy = I_y ** 2
        Ixy = I_x * I_y

        # Step 3: Apply Gaussian integration window to structure tensor components
        Sxx = separable_gaussian_convolve(Ixx, self.sigma)
        Syy = separable_gaussian_convolve(Iyy, self.sigma)
        Sxy = separable_gaussian_convolve(Ixy, self.sigma)

        # Step 4: Compute Corner Response Function (R)
        det_M = (Sxx * Syy) - (Sxy ** 2)
        trace_M = Sxx + Syy
        R = det_M - self.k * (trace_M ** 2)

        # Step 5: Vectorized Non-Maximum Suppression (NMS)
        # Isolate local peaks using a 3x3 max filter
        local_max = ndi.maximum_filter(R, size=3)
        is_max = (R == local_max)

        # Thresholding
        threshold = self.threshold_ratio * R.max() if R.max() > 0 else 0
        corners = is_max & (R > threshold)

        return R, corners, (I_x, I_y)

# Test the detector on a synthetic test pattern (checkerboard)
checkerboard = np.zeros((200, 200), dtype=np.float32)
for i in range(4):
    for j in range(4):
        if (i + j) % 2 == 0:
            checkerboard[i*50:(i+1)*50, j*50:(j+1)*50] = 255.0

# Blur slightly to simulate real conditions
checkerboard_blurred = ndi.gaussian_filter(checkerboard, sigma=1.0)

detector = HarrisCornerDetector(sigma=1.2, k=0.04, threshold_ratio=0.01)
R_map, corner_mask, gradients = detector.detect(checkerboard_blurred)

print(f"Total corners detected on test pattern: {np.sum(corner_mask)}")
```

```python
def visualize_results(image, corners, title="Harris Corner Detection Output"):
    """Visualizes detected corners overlaid on the input image."""
    plt.figure(figsize=(8, 8))
    if len(image.shape) == 2:
        plt.imshow(image, cmap='gray')
    else:
        plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))

    y_coords, x_coords = np.where(corners)
    plt.scatter(x_coords, y_coords, s=25, c='red', marker='o', label='Detected Corners')
    plt.title(title, fontsize=14, fontweight='bold')
    plt.legend(loc='upper right')
    plt.axis('off')
    plt.tight_layout()
    plt.show()

# Run visualization on synthetic checkerboard
visualize_results(checkerboard_blurred, corner_mask, title="Task 5.2: Custom Harris Corner Detector Results")
```

```python
# Convert checkerboard to 8-bit image for OpenCV comparison
img_uint8 = np.uint8(checkerboard_blurred)

# OpenCV Harris Corner execution
opencv_response = cv2.cornerHarris(img_uint8, blockSize=3, ksize=3, k=0.04)
opencv_response = cv2.dilate(opencv_response, None) # Dilate to match NMS style peaks
opencv_corners = opencv_response > 0.01 * opencv_response.max()

# Compare match rate overlap
match_overlap = np.logical_and(corner_mask, opencv_corners).sum()
total_custom = corner_mask.sum()
match_rate = (match_overlap / total_custom) * 100 if total_custom > 0 else 0

print(f"OpenCV vs Custom Implementation Match Rate: {match_rate:.2f}%")
visualize_results(checkerboard_blurred, opencv_corners, title="OpenCV Baseline (cv2.cornerHarris)")
```
